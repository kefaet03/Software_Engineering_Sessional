```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("E:\\He_is_enough03 X UniqoXTech X Dreams\\Click_here\\Academic\\CSE3206\\Software_Engineering_Sessional\\Dataset\\Airlines_quantized.txt")

sns.countplot(x='Delay', data=df)
plt.title('Distribution of Delay (0: No, 1: Yes)')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='Airline', y='Delay', data=df, estimator=lambda x: sum(x)/len(x))
plt.title('Delay Rate by Airline')
plt.ylabel('Delay Rate')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.barplot(x='DayOfWeek', y='Delay', data=df, estimator=lambda x: sum(x)/len(x))
plt.title('Delay Rate by Day of Week')
plt.ylabel('Delay Rate')
plt.show()
```

```python
sns.boxplot(x='Delay', y='Length', data=df)
plt.title('Flight Length vs Delay')
plt.ylabel('Flight Length (minutes)')
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.countplot(x='Airline', hue='Delay', data=df)
plt.title('Flight Count by Airline and Delay')
plt.ylabel('Number of Flights')
plt.xticks(rotation=45)
plt.show()
```
